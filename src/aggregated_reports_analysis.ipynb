```python
import pandas as pd
import glob
import os
import pandas as pd

# Set your target directory (change this to your path)
directory = "../aggregate-reports/"

# Find all CSV files in the directory
csv_files = glob.glob(os.path.join(directory, "*.csv"))

# Read and concatenate them into one DataFrame
all_data = pd.concat([pd.read_csv(file) for file in csv_files], ignore_index=True)

```

```python
# I want all repo_urls that are cited, but not found in our license_exposure procedure
# Using these finding, I want to improve our license finding program
filtered_data = all_data[(all_data['cited'] == 1) & (all_data['origin'] == 'license_exposure') & (all_data['repo_url'].str.contains('github.com'))]
```

```python
filtered_data.head(5)
```

```python
pkg_to_inspect = filtered_data[['pkg_name', 'repo_url']].drop_duplicates().reset_index(drop=True)

# add trailing slash
pkg_to_inspect['repo_url'] = pkg_to_inspect['repo_url'].apply(
    lambda url: url if url.endswith('/') else url + '/'
)
```

```python
pkg_to_inspect.head(10)
```

```python
# Ok, it's we have "dagger" stored in our repo directory.
# Maybe our fingerprints.csv is not right
fp = pd.read_csv('../database/fingerprints.csv')
fp.head()
```

```python
# add trailing slash
fp['repo_url'] = fp['repo_url'].apply(
    lambda url: url if url.endswith('/') else url + '/'
)
```

```python
# Now, cross check pkg_to_inspect and fingerprints
fp_intersect = set()

for r in pkg_to_inspect['repo_url'].unique():
    paths = fp.loc[fp['repo_url'] == r, 'repo_url'].unique()
    fp_intersect.update(paths)

print(fp_intersect)
```

```python
# Collect file paths from pkg_to_inspect
pkg_paths = set()

for r in pkg_to_inspect['repo_url'].unique():
    paths = pkg_to_inspect.loc[pkg_to_inspect['repo_url'] == r, 'repo_url'].unique()
    pkg_paths.update(paths)

# Collect file paths from fp
fp_paths = set()

for r in pkg_to_inspect['repo_url'].unique():
    paths = fp.loc[fp['repo_url'] == r, 'repo_url'].unique()
    fp_paths.update(paths)

# Difference: paths in pkg_to_inspect but not in fp
diff_paths = pkg_paths - fp_paths
diff_paths
```
